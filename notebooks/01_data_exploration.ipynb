```python
import pandas as pd

df_total = pd.read_csv("../data/processed/total_customers.csv")
df_churn_dist = pd.read_csv("../data/processed/churn_distribution.csv")
df_churn_rate = pd.read_csv("../data/processed/churn_rate_overall.csv")
df_by_contract = pd.read_csv("../data/processed/churn_by_contract.csv")
df_by_tenure = pd.read_csv("../data/processed/churn_by_tenure_group.csv")
df_by_internet = pd.read_csv("../data/processed/churn_by_internet_service.csv")
df_by_billing = pd.read_csv("../data/processed/churn_by_billing.csv")
df_by_senior = pd.read_csv("../data/processed/churn_by_senior_citizen.csv")
df_summary = pd.read_csv("../data/processed/churn_summary_dashboard.csv")

df_total, df_churn_rate
```

```python
def normalize_columns(df):
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
        .str.replace("%", "percentage")
    )
    return df

dfs = [
    df_churn_dist,
    df_by_contract,
    df_by_tenure,
    df_by_internet,
    df_by_billing,
    df_by_senior,
    df_summary
]

dfs = [normalize_columns(df) for df in dfs]

df_churn_dist.head()
```

```python
kpi_overview = pd.DataFrame({
    "total_customers": [df_total.iloc[0, 0]],
    "churn_rate_percentage": [df_churn_rate.iloc[0, 0]]
})

kpi_overview
```

```python
kpi_overview.to_csv(
    "../data/processed/kpi_overview.csv",
    index=False
)
```

```python
df_churn_dist["churn_label"] = df_churn_dist["churn"].map({
    True: "Churn",
    False: "No Churn"
})

df_churn_dist.to_csv(
    "../data/processed/dashboard_churn_distribution.csv",
    index=False
)
```
